# Stage 2 Analysis

Load the Stage 2 evaluation rollout generated by `rl/scripts/eval_exo_stage2.py` and summarize walker-only versus walker-plus-exo performance.

```python
from pathlib import Path

import numpy as np
import pandas as pd

rollout_path = Path("../rl_output/stage2_eval_rollout.npz")
data = np.load(rollout_path, allow_pickle=True)
df = pd.DataFrame({key: data[key] for key in data.files})
df.head()
```

```python
episode_lengths = (
    df.groupby(["condition", "episode"]).size().rename("episode_length").reset_index()
)

summary = df.groupby("condition").agg(
    mean_reward=("reward", "mean"),
    mean_effort=("effort", "mean"),
    mean_torque=("mean_torque", "mean"),
)
summary = summary.join(
    episode_lengths.groupby("condition")["episode_length"].mean().rename("mean_episode_length")
)
summary
```
